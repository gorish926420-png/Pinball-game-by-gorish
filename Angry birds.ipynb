```python
from IPython.display import HTML, display

display(HTML(r"""
<div id="carrom-game" style="max-width:660px;margin:12px auto;font-family:Arial,sans-serif">
  <div style="display:flex;justify-content:space-between;align-items:center;margin-bottom:8px">
    <strong style="font-size:20px">CARROM</strong>
    <span id="carrom-status">Player 1's turn</span>
    <button id="carrom-reset" style="padding:7px 12px;cursor:pointer">New game</button>
  </div>
  <canvas id="carrom-canvas" width="640" height="640"
          style="width:100%;height:auto;touch-action:none;border-radius:12px;cursor:crosshair"></canvas>
  <div style="font-size:13px;color:#555;margin-top:6px">
    Drag the striker backward to aim and release to shoot. Pocket a coin to score.
  </div>
</div>

<script>
(() => {
  const canvas = document.getElementById("carrom-canvas");
  const ctx = canvas.getContext("2d");
  const status = document.getElementById("carrom-status");
  const resetButton = document.getElementById("carrom-reset");

  const W = canvas.width, H = canvas.height;
  const cx = W / 2, cy = H / 2;
  const pocketR = 25, coinR = 13, strikerR = 17;
  const friction = 0.992, wallBounce = 0.82;
  const minSpeed = 0.045, maxShot = 15;
  const board = { left: 55, right: 585, top: 55, bottom: 585 };

  let pieces = [], striker, player = 1, scores = [0, 0];
  let dragging = false, aimX = 0, aimY = 0, gameOver = false;
  let shotTaken = false;

  function makePiece(x, y, color, kind="coin") {
    return { x, y, vx: 0, vy: 0, r: kind === "striker" ? strikerR : coinR,
             color, kind, active: true };
  }

  function newGame() {
    pieces = [];
    // Center cluster: queen, then alternating light and dark coins.
    pieces.push(makePiece(cx, cy, "#d7263d", "queen"));
    const colors = ["#f5d76e", "#273043"];
    const ring = [
      [0,-29], [25,-14], [25,14], [0,29], [-25,14], [-25,-14],
      [0,-58], [50,-29], [50,29], [0,58], [-50,29], [-50,-29]
    ];
    ring.forEach((p, i) => pieces.push(makePiece(cx+p[0], cy+p[1], colors[i % 2])));
    scores = [0, 0];
    player = 1;
    gameOver = false;
    shotTaken = false;
    placeStriker();
    updateStatus();
    draw();
  }

  function placeStriker() {
    striker = makePiece(cx, player === 1 ? 520 : 120, "#3a86ff", "striker");
    pieces.push(striker);
  }

  function updateStatus() {
    status.textContent = gameOver
      ? `Player ${scores[0] >= 5 ? 1 : 2} wins! (${scores[0]}–${scores[1]})`
      : `Player ${player}'s turn  ·  Score ${scores[0]}–${scores[1]}`;
  }

  function allStopped() {
    return pieces.every(p => !p.active || Math.hypot(p.vx, p.vy) < minSpeed);
  }

  function nextTurn() {
    if (scores[player - 1] >= 5) {
      gameOver = true;
      updateStatus();
      return;
    }
    player = player === 1 ? 2 : 1;
    shotTaken = false;
    placeStriker();
    updateStatus();
  }

  function shoot(dx, dy) {
    const length = Math.hypot(dx, dy);
    if (length < 6) return;
    const power = Math.min(length * 0.12, maxShot);
    striker.vx = (dx / length) * power;
    striker.vy = (dy / length) * power;
    shotTaken = true;
    dragging = false;
  }

  function pocket(piece) {
    piece.active = false;
    piece.vx = piece.vy = 0;
    if (piece.kind === "coin" || piece.kind === "queen") {
      scores[player - 1]++;
      updateStatus();
    }
    // A pocketed striker is returned to its baseline after the shot.
  }

  function update() {
    if (!gameOver) {
      for (const p of pieces) {
        if (!p.active) continue;
        p.x += p.vx;
        p.y += p.vy;
        p.vx *= friction;
        p.vy *= friction;
        if (Math.hypot(p.vx, p.vy) < minSpeed) p.vx = p.vy = 0;

        // Pocket check
        const pockets = [[board.left,board.top], [board.right,board.top],
                         [board.left,board.bottom], [board.right,board.bottom]];
        if (pockets.some(([x,y]) => Math.hypot(p.x-x, p.y-y) < pocketR)) {
          pocket(p);
          continue;
        }

        // Cushions
        if (p.x - p.r < board.left) {
          p.x = board.left + p.r; p.vx = Math.abs(p.vx) * wallBounce;
        }
        if (p.x + p.r > board.right) {
          p.x = board.right - p.r; p.vx = -Math.abs(p.vx) * wallBounce;
        }
        if (p.y - p.r < board.top) {
          p.y = board.top + p.r; p.vy = Math.abs(p.vy) * wallBounce;
        }
        if (p.y + p.r > board.bottom) {
          p.y = board.bottom - p.r; p.vy = -Math.abs(p.vy) * wallBounce;
        }
      }

      // Circle collisions
      for (let i = 0; i < pieces.length; i++) {
        const a = pieces[i];
        if (!a.active) continue;
        for (let j = i + 1; j < pieces.length; j++) {
          const b = pieces[j];
          if (!b.active) continue;
          let dx = b.x-a.x, dy = b.y-a.y;
          let d = Math.hypot(dx,dy), minD = a.r+b.r;
          if (d === 0) { dx = 0.01; d = 0.01; }
          if (d < minD) {
            const nx = dx/d, ny = dy/d;
            const overlap = minD-d;
            a.x -= nx*overlap/2; a.y -= ny*overlap/2;
            b.x += nx*overlap/2; b.y += ny*overlap/2;
            const rel = (a.vx-b.vx)*nx + (a.vy-b.vy)*ny;
            if (rel > 0) {
              const impulse = rel * 0.92;
              a.vx -= impulse*nx; a.vy -= impulse*ny;
              b.vx += impulse*nx; b.vy += impulse*ny;
            }
          }
        }
      }

      if (shotTaken && allStopped()) {
        pieces = pieces.filter(p => p !== striker);
        nextTurn();
      }
    }
    draw();
    requestAnimationFrame(update);
  }

  function draw() {
    ctx.clearRect(0, 0, W, H);
    ctx.fillStyle = "#7a4827";
    ctx.fillRect(0, 0, W, H);
    ctx.fillStyle = "#e8c994";
    ctx.fillRect(board.left, board.top, board.right-board.left, board.bottom-board.top);

    // Board border and baseline
    ctx.strokeStyle = "#6b3e20";
    ctx.lineWidth = 8;
    ctx.strokeRect(board.left, board.top, board.right-board.left, board.bottom-board.top);
    ctx.lineWidth = 3;
    ctx.strokeStyle = "#a65b35";
    [120, 520].forEach(y => {
      ctx.beginPath(); ctx.moveTo(135,y); ctx.lineTo(505,y); ctx.stroke();
      [135,505].forEach(x => {
        ctx.beginPath(); ctx.arc(x,y,8,0,Math.PI*2); ctx.stroke();
      });
    });

    // Center design
    ctx.strokeStyle = "#a65b35";
    ctx.lineWidth = 2;
    ctx.beginPath(); ctx.arc(cx,cy,62,0,Math.PI*2); ctx.stroke();
    ctx.beginPath(); ctx.arc(cx,cy,36,0,Math.PI*2); ctx.stroke();
    ctx.fillStyle = "#d7263d";
    ctx.beginPath(); ctx.arc(cx,cy,5,0,Math.PI*2); ctx.fill();

    // Pockets
    [[board.left,board.top], [board.right,board.top],
     [board.left,board.bottom], [board.right,board.bottom]].forEach(([x,y]) => {
      ctx.fillStyle = "#30231c";
      ctx.beginPath(); ctx.arc(x,y,pocketR,0,Math.PI*2); ctx.fill();
      ctx.strokeStyle = "#c78d54"; ctx.lineWidth = 3;
      ctx.stroke();
    });

    for (const p of pieces) {
      if (!p.active) continue;
      ctx.fillStyle = "rgba(40,25,15,.22)";
      ctx.beginPath(); ctx.arc(p.x+2,p.y+3,p.r,0,Math.PI*2); ctx.fill();
      ctx.fillStyle = p.color;
      ctx.beginPath(); ctx.arc(p.x,p.y,p.r,0,Math.PI*2); ctx.fill();
      ctx.strokeStyle = p.kind === "striker" ? "#174a9c" : "#805b36";
      ctx.lineWidth = 2;
      ctx.stroke();
      ctx.fillStyle = "rgba(255,255,255,.35)";
      ctx.beginPath(); ctx.arc(p.x-p.r*.3,p.y-p.r*.3,p.r*.22,0,Math.PI*2); ctx.fill();
    }

    if (dragging) {
      ctx.strokeStyle = "#1769aa";
      ctx.lineWidth = 3;
      ctx.setLineDash([7,5]);
      ctx.beginPath(); ctx.moveTo(striker.x,striker.y);
      ctx.lineTo(striker.x + (striker.x-aimX), striker.y + (striker.y-aimY));
      ctx.stroke();
      ctx.setLineDash([]);
    }
  }

  function pointerPos(event) {
    const rect = canvas.getBoundingClientRect();
    return { x: (event.clientX-rect.left)*W/rect.width,
             y: (event.clientY-rect.top)*H/rect.height };
  }

  canvas.addEventListener("pointerdown", event => {
    if (gameOver || shotTaken || !allStopped()) return;
    const p = pointerPos(event);
    if (Math.hypot(p.x-striker.x,p.y-striker.y) <= striker.r+10) {
      dragging = true; aimX = p.x; aimY = p.y;
      canvas.setPointerCapture(event.pointerId);
    }
  });
  canvas.addEventListener("pointermove", event => {
    if (dragging) { const p=pointerPos(event); aimX=p.x; aimY=p.y; draw(); }
  });
  canvas.addEventListener("pointerup", event => {
    if (!dragging) return;
    const p=pointerPos(event);
    shoot(striker.x-p.x, striker.y-p.y);
  });
  resetButton.addEventListener("click", newGame);

  newGame();
  requestAnimationFrame(update);
})();
</script>
"""))
```

```python
import random
import time
from IPython.display import clear_output

level = 1
print("🧠 NUMBER MEMORY GAME")
print("Memorize each number, then type it back.")
input("Press Enter to start...")

while True:
    # Show a longer number as the level increases
    digits = min(level + 2, 12)
    number = "".join(str(random.randint(0, 9)) for _ in range(digits))

    clear_output(wait=True)
    print(f"🧠 Level {level}")
    print(f"\nMemorize this number:\n\n   {number}")
    time.sleep(max(1.5, digits * 0.7))

    clear_output(wait=True)
    answer = input(f"Level {level}: Type the number you memorized: ")

    if answer == number:
        level += 1
        print("✅ Correct! Get ready for the next level...")
        time.sleep(1.2)
    else:
        print(f"❌ Not quite. The number was {number}.")
        print(f"You reached level {level}.")
        break
```

```python
import random
import time
from IPython.display import clear_output

directions = ["left", "center", "right"]
score = 0
shots = 5

print("⚽ PENALTY SHOOTOUT")
print("You have 5 shots. Pick left, center, or right.")
input("Press Enter to start...")

for shot in range(1, shots + 1):
    clear_output(wait=True)
    print(f"⚽ Shot {shot} of {shots}")
    print(f"Your score: {score}")
    choice = input("Shoot left, center, or right? ").strip().lower()

    while choice not in directions:
        choice = input("Please type left, center, or right: ").strip().lower()

    goalkeeper = random.choice(directions)
    print(f"\nYou shoot {choice}...")
    time.sleep(0.8)
    print(f"The goalkeeper dives {goalkeeper}!")

    if choice == goalkeeper:
        print("🧤 Saved!")
    else:
        score += 1
        print("🥅 GOAL!")

    time.sleep(1.5)

clear_output(wait=True)
print("🏁 FULL TIME")
print(f"You scored {score} out of {shots}.")

if score == shots:
    print("🏆 Perfect shootout!")
elif score >= 3:
    print("👏 Great shooting!")
else:
    print("⚽ Good try! Play again and beat your score.")
```

```python
from IPython.display import HTML, display

display(HTML(r"""
<div style="max-width:500px;margin:12px auto;font-family:Arial,sans-serif">
  <div style="display:flex;justify-content:space-between;align-items:center;margin-bottom:8px">
    <strong style="font-size:20px">PINBALL</strong>
    <span id="pinball-score">Score: 0 &nbsp; Balls: 3</span>
    <button id="pinball-reset" style="padding:6px 10px;cursor:pointer">New game</button>
  </div>
  <canvas id="pinball-board" width="480" height="680"
          tabindex="0"
          style="width:100%;height:auto;background:#10182a;border:4px solid #343f57;border-radius:12px;display:block;outline:none"></canvas>
  <div style="font-size:13px;color:#555;margin-top:6px">
    Click the board, then hold ← and → to control the flippers. Press Space to launch.
  </div>
</div>

<script>
(() => {
  const canvas = document.getElementById("pinball-board");
  const ctx = canvas.getContext("2d");
  const scoreText = document.getElementById("pinball-score");
  const resetButton = document.getElementById("pinball-reset");
  const W = canvas.width, H = canvas.height;

  let ball, flippers, score, lives, keys, running, gameOver;
  const bumpers = [
    {x:150, y:190, r:32, points:100},
    {x:320, y:190, r:32, points:100},
    {x:235, y:295, r:38, points:250},
    {x:115, y:365, r:23, points:150},
    {x:355, y:365, r:23, points:150}
  ];
  const walls = [
    {x1:45,y1:35,x2:435,y2:35},
    {x1:45,y1:35,x2:45,y2:570},
    {x1:435,y1:35,x2:435,y2:570},
    {x1:45,y1:570,x2:165,y2:645},
    {x1:435,y1:570,x2:315,y2:645}
  ];

  function resetGame() {
    score = 0; lives = 3; gameOver = false;
    keys = {left:false,right:false};
    flippers = {left:false,right:false};
    launchBall();
    updateScore();
    canvas.focus();
  }

  function launchBall() {
    ball = {x:405, y:580, vx:0, vy:0, r:9, launched:false};
  }

  function updateScore() {
    scoreText.textContent = `Score: ${score}   Balls: ${lives}`;
  }

  function launch() {
    if (!ball.launched && !gameOver) {
      ball.launched = true;
      ball.vx = -2.2;
      ball.vy = -10.5;
    }
  }

  function keyDown(e) {
    if (["ArrowLeft","ArrowRight","Space"].includes(e.code)) e.preventDefault();
    if (e.code === "ArrowLeft") keys.left = true;
    if (e.code === "ArrowRight") keys.right = true;
    if (e.code === "Space") launch();
  }
  function keyUp(e) {
    if (e.code === "ArrowLeft") keys.left = false;
    if (e.code === "ArrowRight") keys.right = false;
  }

  canvas.addEventListener("keydown", keyDown);
  canvas.addEventListener("keyup", keyUp);
  window.addEventListener("keydown", keyDown);
  window.addEventListener("keyup", keyUp);
  resetButton.addEventListener("click", resetGame);
  canvas.addEventListener("pointerdown", () => canvas.focus());

  function collideCircle(cx, cy, radius, points) {
    const dx = ball.x-cx, dy = ball.y-cy;
    const d = Math.hypot(dx,dy), min = ball.r+radius;
    if (d < min) {
      const nx = d ? dx/d : 0, ny = d ? dy/d : -1;
      ball.x = cx + nx*min;
      ball.y = cy + ny*min;
      const dot = ball.vx*nx + ball.vy*ny;
      if (dot < 0) {
        ball.vx -= 1.85*dot*nx;
        ball.vy -= 1.85*dot*ny;
      }
      ball.vx += nx*2.0;
      ball.vy += ny*2.0;
      score += points;
      updateScore();
    }
  }

  function collideSegment(x1,y1,x2,y2, restitution=0.78) {
    const sx=x2-x1, sy=y2-y1;
    const length2=sx*sx+sy*sy;
    let t=((ball.x-x1)*sx+(ball.y-y1)*sy)/length2;
    t=Math.max(0,Math.min(1,t));
    const px=x1+t*sx, py=y1+t*sy;
    let dx=ball.x-px, dy=ball.y-py;
    const d=Math.hypot(dx,dy);
    if (d < ball.r) {
      let nx=d ? dx/d : -sy/Math.hypot(sx,sy);
      let ny=d ? dy/d : sx/Math.hypot(sx,sy);
      ball.x=px+nx*ball.r;
      ball.y=py+ny*ball.r;
      const dot=ball.vx*nx+ball.vy*ny;
      if(dot<0) {
        ball.vx-=(1+restitution)*dot*nx;
        ball.vy-=(1+restitution)*dot*ny;
      }
    }
  }

  function update() {
    if (!gameOver && ball.launched) {
      ball.vy += 0.20;
      ball.x += ball.vx;
      ball.y += ball.vy;
      ball.vx *= 0.999;
      ball.vy *= 0.999;

      for (const w of walls) collideSegment(w.x1,w.y1,w.x2,w.y2);

      for (const b of bumpers) collideCircle(b.x,b.y,b.r,b.points);

      // Flippers snap up when their arrow key is held.
      const leftTip = keys.left ? {x:220,y:530} : {x:205,y:570};
      const rightTip = keys.right ? {x:260,y:530} : {x:275,y:570};
      collideSegment(130,585,leftTip.x,leftTip.y,0.9);
      collideSegment(350,585,rightTip.x,rightTip.y,0.9);

      if (ball.y > H+30 || ball.x < -30 || ball.x > W+30) {
        lives--;
        updateScore();
        if (lives <= 0) gameOver = true;
        else launchBall();
      }
    }
    draw();
    requestAnimationFrame(update);
  }

  function draw() {
    ctx.clearRect(0,0,W,H);

    // Playfield and lane
    ctx.fillStyle="#10182a"; ctx.fillRect(0,0,W,H);
    ctx.strokeStyle="#74819b"; ctx.lineWidth=5;
    for (const w of walls) {
      ctx.beginPath(); ctx.moveTo(w.x1,w.y1); ctx.lineTo(w.x2,w.y2); ctx.stroke();
    }
    ctx.strokeStyle="#4b5872"; ctx.lineWidth=2;
    ctx.beginPath(); ctx.moveTo(390,50); ctx.lineTo(390,590); ctx.stroke();
    ctx.fillStyle="#26344d"; ctx.fillRect(390,535,45,55);
    ctx.fillStyle="#a9b8d0"; ctx.font="13px Arial"; ctx.fillText("SPACE",393,565);

    // Bumpers
    for (const b of bumpers) {
      const grad=ctx.createRadialGradient(b.x-8,b.y-8,4,b.x,b.y,b.r);
      grad.addColorStop(0,"#fff1a8"); grad.addColorStop(.35,"#f9a826");
      grad.addColorStop(1,"#d34f24");
      ctx.fillStyle=grad; ctx.beginPath(); ctx.arc(b.x,b.y,b.r,0,Math.PI*2); ctx.fill();
      ctx.strokeStyle="#ffe08a"; ctx.lineWidth=4; ctx.stroke();
      ctx.fillStyle="#fff7d0"; ctx.font="bold 13px Arial";
      ctx.textAlign="center"; ctx.fillText("+"+b.points,b.x,b.y+5);
      ctx.textAlign="left";
    }

    // Flippers
    const lt=keys.left ? {x:220,y:530} : {x:205,y:570};
    const rt=keys.right ? {x:260,y:530} : {x:275,y:570};
    ctx.lineCap="round"; ctx.lineWidth=17; ctx.strokeStyle="#53d6c5";
    ctx.beginPath(); ctx.moveTo(130,585); ctx.lineTo(lt.x,lt.y); ctx.stroke();
    ctx.beginPath(); ctx.moveTo(350,585); ctx.lineTo(rt.x,rt.y); ctx.stroke();
    ctx.fillStyle="#d8fff8";
    for(const p of [{x:130,y:585},{x:350,y:585}]) {
      ctx.beginPath(); ctx.arc(p.x,p.y,11,0,Math.PI*2); ctx.fill();
    }

    // Ball
    const ballGrad=ctx.createRadialGradient(ball.x-3,ball.y-4,1,ball.x,ball.y,ball.r);
    ballGrad.addColorStop(0,"#fff"); ballGrad.addColorStop(1,"#a9c6ff");
    ctx.fillStyle=ballGrad; ctx.beginPath(); ctx.arc(ball.x,ball.y,ball.r,0,Math.PI*2); ctx.fill();

    if (!ball.launched && !gameOver) {
      ctx.fillStyle="#d9e2f2"; ctx.font="bold 15px Arial";
      ctx.textAlign="center"; ctx.fillText("Press SPACE to launch",240,625);
      ctx.textAlign="left";
    }
    if (gameOver) {
      ctx.fillStyle="rgba(5,10,20,.78)"; ctx.fillRect(45,270,390,110);
      ctx.fillStyle="#fff"; ctx.textAlign="center"; ctx.font="bold 28px Arial";
      ctx.fillText("GAME OVER",240,315);
      ctx.font="18px Arial"; ctx.fillText(`Final score: ${score}`,240,350);
      ctx.textAlign="left";
    }
  }

  resetGame();
  requestAnimationFrame(update);
})();
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML(r"""
<div style="max-width:560px;margin:12px auto;font-family:Arial,sans-serif">
  <div style="display:flex;justify-content:space-between;align-items:center;margin-bottom:8px">
    <strong style="font-size:20px">SNAKE &amp; BALL</strong>
    <span id="snake-score">Score: 0</span>
    <button id="snake-restart" style="padding:6px 10px;cursor:pointer">Restart</button>
  </div>
  <canvas id="snake-board" width="500" height="500" tabindex="0"
          style="width:100%;height:auto;background:#101b2d;border:3px solid #33445f;border-radius:10px;display:block;outline:none"></canvas>
  <div style="font-size:13px;color:#555;margin-top:6px">
    Click the board first. Move with arrow keys or WASD. Press Space to pause.
  </div>
</div>

<script>
(() => {
  const canvas = document.getElementById("snake-board");
  const ctx = canvas.getContext("2d");
  const scoreText = document.getElementById("snake-score");
  const restartButton = document.getElementById("snake-restart");

  const cell = 20, cols = 25, rows = 25;
  let snake, direction, nextDirection, ball, score, paused, over;
  let lastStep = 0, stepMs = 115;

  function reset() {
    snake = [{x:12,y:12},{x:11,y:12},{x:10,y:12}];
    direction = {x:1,y:0};
    nextDirection = {x:1,y:0};
    score = 0;
    paused = false;
    over = false;
    stepMs = 115;
    lastStep = 0;
    placeBall();
    updateScore();
    draw();
    canvas.focus();
  }

  function placeBall() {
    const free = [];
    for (let y=0; y<rows; y++) {
      for (let x=0; x<cols; x++) {
        if (!snake.some(part => part.x===x && part.y===y)) free.push({x,y});
      }
    }
    ball = free.length ? free[Math.floor(Math.random()*free.length)] : null;
  }

  function updateScore() {
    scoreText.textContent = `Score: ${score}`;
  }

  function handleKey(e) {
    const key = e.key.toLowerCase();
    if (["arrowup","arrowdown","arrowleft","arrowright"," "].includes(key)) e.preventDefault();

    if (key === " " && !over) {
      paused = !paused;
      draw();
      return;
    }
    if (key === "r") {
      reset();
      return;
    }

    const moves = {
      arrowup:{x:0,y:-1}, w:{x:0,y:-1},
      arrowdown:{x:0,y:1}, s:{x:0,y:1},
      arrowleft:{x:-1,y:0}, a:{x:-1,y:0},
      arrowright:{x:1,y:0}, d:{x:1,y:0}
    };
    const move = moves[key];
    if (move && !(move.x === -direction.x && move.y === -direction.y)) {
      nextDirection = move;
    }
  }

  window.addEventListener("keydown", handleKey);
  canvas.addEventListener("pointerdown", () => canvas.focus());
  restartButton.addEventListener("click", reset);

  function step() {
    if (paused || over) return;

    direction = nextDirection;
    const head = {x:snake[0].x+direction.x, y:snake[0].y+direction.y};
    const eating = ball && head.x===ball.x && head.y===ball.y;
    const bodyToCheck = eating ? snake : snake.slice(0,-1);

    if (head.x<0 || head.x>=cols || head.y<0 || head.y>=rows ||
        bodyToCheck.some(part => part.x===head.x && part.y===head.y)) {
      over = true;
      return;
    }

    snake.unshift(head);
    if (eating) {
      score++;
      updateScore();
      stepMs = Math.max(55, 115-score*2);
      placeBall();
      if (!ball) over = true;
    } else {
      snake.pop();
    }
  }

  function draw() {
    ctx.clearRect(0,0,canvas.width,canvas.height);

    // Grid
    ctx.strokeStyle="#1a2a42";
    ctx.lineWidth=1;
    for (let i=0; i<=cols; i++) {
      ctx.beginPath(); ctx.moveTo(i*cell,0); ctx.lineTo(i*cell,canvas.height); ctx.stroke();
    }
    for (let i=0; i<=rows; i++) {
      ctx.beginPath(); ctx.moveTo(0,i*cell); ctx.lineTo(canvas.width,i*cell); ctx.stroke();
    }

    // Ball
    if (ball) {
      ctx.fillStyle="#fb5267";
      ctx.beginPath(); ctx.arc(ball.x*cell+cell/2,ball.y*cell+cell/2,8,0,Math.PI*2); ctx.fill();
      ctx.fillStyle="#ffd8de";
      ctx.beginPath(); ctx.arc(ball.x*cell+7,ball.y*cell+6,2,0,Math.PI*2); ctx.fill();
    }

    // Snake
    snake.forEach((part,i) => {
      ctx.fillStyle = i===0 ? "#61e6a5" : "#22bd7b";
      ctx.beginPath();
      ctx.roundRect(part.x*cell+2,part.y*cell+2,cell-4,cell-4,5);
      ctx.fill();
    });

    if (paused || over) {
      ctx.fillStyle="rgba(5,10,20,.78)";
      ctx.fillRect(0,205,canvas.width,90);
      ctx.fillStyle="#fff";
      ctx.textAlign="center";
      ctx.font="bold 26px Arial";
      ctx.fillText(over ? "GAME OVER" : "PAUSED",canvas.width/2,244);
      ctx.font="15px Arial";
      ctx.fillText(over ? "Press R or Restart to play again" : "Press Space to continue",canvas.width/2,271);
      ctx.textAlign="left";
    }
  }

  function loop(timestamp) {
    if (!lastStep) lastStep = timestamp;
    if (timestamp-lastStep >= stepMs) {
      step();
      draw();
      lastStep = timestamp;
    }
    requestAnimationFrame(loop);
  }

  reset();
  requestAnimationFrame(loop);
})();
</script>
"""))
```

```python
from IPython.display import HTML, display

display(HTML(r"""
<div style="max-width:760px;margin:12px auto;font-family:Arial,sans-serif">
  <div style="display:flex;justify-content:space-between;align-items:center;margin-bottom:8px">
    <strong style="font-size:20px">SLINGSHOT BIRDS</strong>
    <span id="bird-score">Targets: 0 / 3</span>
    <button id="bird-reset" style="padding:6px 10px;cursor:pointer">Reset</button>
  </div>
  <canvas id="bird-game" width="720" height="430" tabindex="0"
          style="width:100%;height:auto;display:block;background:#bde8ff;border:3px solid #567;border-radius:10px;outline:none;touch-action:none"></canvas>
  <div style="font-size:13px;color:#555;margin-top:6px">
    Drag the red bird back from the slingshot and release. Press R to reset.
  </div>
</div>

<script>
(() => {
  const canvas = document.getElementById("bird-game");
  const ctx = canvas.getContext("2d");
  const scoreLabel = document.getElementById("bird-score");
  const resetButton = document.getElementById("bird-reset");
  const W = canvas.width, H = canvas.height;
  const anchor = {x:145, y:285};
  const groundY = 365;
  const birdRadius = 16;

  let bird, blocks, pigs, dragging, flying, shots, score, settledFrames;

  function reset() {
    bird = {x:anchor.x, y:anchor.y, vx:0, vy:0, r:birdRadius};
    blocks = [
      {x:475,y:315,w:22,h:50,hp:1},
      {x:555,y:315,w:22,h:50,hp:1},
      {x:485,y:278,w:82,h:18,hp:1},
      {x:505,y:245,w:22,h:48,hp:1},
      {x:545,y:245,w:22,h:48,hp:1},
      {x:505,y:211,w:62,h:18,hp:1}
    ];
    pigs = [
      {x:530,y:337,r:15,alive:true},
      {x:535,y:190,r:15,alive:true},
      {x:610,y:345,r:15,alive:true}
    ];
    dragging = false;
    flying = false;
    shots = 0;
    settledFrames = 0;
    updateScore();
    draw();
  }

  function updateScore() {
    const hit = pigs.filter(p => !p.alive).length;
    scoreLabel.textContent = `Targets: ${hit} / ${pigs.length} · Shots: ${shots}`;
  }

  function pointerPos(e) {
    const r = canvas.getBoundingClientRect();
    return {x:(e.clientX-r.left)*W/r.width, y:(e.clientY-r.top)*H/r.height};
  }

  canvas.addEventListener("pointerdown", e => {
    const p = pointerPos(e);
    if (!flying && bird.x < 260 && Math.hypot(p.x-bird.x,p.y-bird.y) < 35) {
      dragging = true;
      canvas.setPointerCapture(e.pointerId);
    }
  });

  canvas.addEventListener("pointermove", e => {
    if (!dragging) return;
    const p = pointerPos(e);
    const dx = p.x-anchor.x, dy = p.y-anchor.y;
    const maxPull = 85, dist = Math.hypot(dx,dy);
    const scale = dist > maxPull ? maxPull/dist : 1;
    bird.x = anchor.x + dx*scale;
    bird.y = anchor.y + dy*scale;
    draw();
  });

  canvas.addEventListener("pointerup", () => {
    if (!dragging) return;
    dragging = false;
    const power = 0.20;
    bird.vx = (anchor.x-bird.x)*power;
    bird.vy = (anchor.y-bird.y)*power;
    flying = true;
    shots++;
    updateScore();
  });

  function circleRectHit(c, r) {
    const px = Math.max(r.x, Math.min(c.x, r.x+r.w));
    const py = Math.max(r.y, Math.min(c.y, r.y+r.h));
    const dx=c.x-px, dy=c.y-py;
    return dx*dx+dy*dy < c.r*c.r;
  }

  function update() {
    if (flying) {
      bird.vy += 0.32;
      bird.x += bird.vx;
      bird.y += bird.vy;
      bird.vx *= 0.995;
      bird.vy *= 0.995;

      // Ground bounce
      if (bird.y+bird.r > groundY) {
        bird.y = groundY-bird.r;
        bird.vy *= -0.48;
        bird.vx *= 0.78;
      }

      // Hit blocks
      for (const b of blocks) {
        if (b.hp > 0 && circleRectHit(bird,b)) {
          b.hp = 0;
          bird.vx *= -0.48;
          bird.vy *= 0.65;
        }
      }

      // Hit targets
      for (const p of pigs) {
        if (p.alive && Math.hypot(bird.x-p.x,bird.y-p.y) < bird.r+p.r) {
          p.alive = false;
          bird.vx *= 0.72;
          bird.vy *= 0.72;
          updateScore();
        }
      }

      const speed = Math.hypot(bird.vx,bird.vy);
      if (speed < 1.1) settledFrames++; else settledFrames=0;
      if (bird.x > W+80 || bird.y > H+100 || settledFrames > 45) {
        flying = false;
        if (shots < 3 && pigs.some(p => p.alive)) {
          bird = {x:anchor.x,y:anchor.y,vx:0,vy:0,r:birdRadius};
        }
      }
    }
    draw();
    requestAnimationFrame(update);
  }

  function draw() {
    ctx.clearRect(0,0,W,H);

    // Sky and ground
    const sky = ctx.createLinearGradient(0,0,0,H);
    sky.addColorStop(0,"#9ee3ff"); sky.addColorStop(1,"#e8f8ff");
    ctx.fillStyle=sky; ctx.fillRect(0,0,W,H);
    ctx.fillStyle="#82c95b"; ctx.fillRect(0,groundY,W,H-groundY);
    ctx.fillStyle="#5b9f3f"; ctx.fillRect(0,groundY,W,8);

    // Clouds
    ctx.fillStyle="rgba(255,255,255,.8)";
    [[95,75],[330,105],[600,70]].forEach(([x,y]) => {
      ctx.beginPath(); ctx.arc(x,y,20,0,Math.PI*2);
      ctx.arc(x+22,y-8,25,0,Math.PI*2);
      ctx.arc(x+47,y,19,0,Math.PI*2); ctx.fill();
    });

    // Slingshot
    ctx.strokeStyle="#70421f"; ctx.lineWidth=12; ctx.lineCap="round";
    ctx.beginPath(); ctx.moveTo(125,groundY); ctx.lineTo(145,285);
    ctx.moveTo(164,groundY); ctx.lineTo(145,285); ctx.stroke();
    if (dragging || !flying) {
      ctx.strokeStyle="#5b3522"; ctx.lineWidth=5;
      ctx.beginPath(); ctx.moveTo(135,278); ctx.lineTo(bird.x,bird.y);
      ctx.stroke();
    }

    // Blocks
    for (const b of blocks) {
      if (b.hp <= 0) continue;
      ctx.fillStyle="#c98b52";
      ctx.fillRect(b.x,b.y,b.w,b.h);
      ctx.strokeStyle="#81552e"; ctx.lineWidth=3;
      ctx.strokeRect(b.x,b.y,b.w,b.h);
      ctx.beginPath();
      ctx.moveTo(b.x+4,b.y+4); ctx.lineTo(b.x+b.w-4,b.y+b.h-4);
      ctx.moveTo(b.x+b.w-4,b.y+4); ctx.lineTo(b.x+4,b.y+b.h-4);
      ctx.stroke();
    }

    // Targets
    for (const p of pigs) {
      if (!p.alive) continue;
      ctx.fillStyle="#65c94a";
      ctx.beginPath(); ctx.arc(p.x,p.y,p.r,0,Math.PI*2); ctx.fill();
      ctx.strokeStyle="#347d2c"; ctx.lineWidth=2; ctx.stroke();
      ctx.fillStyle="#fff";
      ctx.beginPath(); ctx.arc(p.x-5,p.y-3,4,0,Math.PI*2);
      ctx.arc(p.x+5,p.y-3,4,0,Math.PI*2); ctx.fill();
      ctx.fillStyle="#222";
      ctx.beginPath(); ctx.arc(p.x-4,p.y-3,2,0,Math.PI*2);
      ctx.arc(p.x+6,p.y-3,2,0,Math.PI*2); ctx.fill();
      ctx.strokeStyle="#347d2c";
      ctx.beginPath(); ctx.arc(p.x,p.y+5,5,0,Math.PI); ctx.stroke();
    }

    // Bird
    ctx.fillStyle="#e53935";
    ctx.beginPath(); ctx.arc(bird.x,bird.y,bird.r,0,Math.PI*2); ctx.fill();
    ctx.strokeStyle="#9c1c1c"; ctx.lineWidth=2; ctx.stroke();
    ctx.fillStyle="#fff";
    ctx.beginPath(); ctx.arc(bird.x+5,bird.y-5,5,0,Math.PI*2); ctx.fill();
    ctx.fillStyle="#222";
    ctx.beginPath(); ctx.arc(bird.x+7,bird.y-5,2,0,Math.PI*2); ctx.fill();
    ctx.fillStyle="#ffca28";
    ctx.beginPath(); ctx.moveTo(bird.x+12,bird.y);
    ctx.lineTo(bird.x+24,bird.y+4); ctx.lineTo(bird.x+12,bird.y+7); ctx.fill();

    if (pigs.every(p => !p.alive)) {
      ctx.fillStyle="rgba(0,0,0,.55)"; ctx.fillRect(180,145,360,70);
      ctx.fillStyle="#fff"; ctx.font="bold 28px Arial"; ctx.textAlign="center";
      ctx.fillText("All targets cleared!",360,190); ctx.textAlign="left";
    } else if (shots >= 3 && !flying && pigs.some(p => p.alive)) {
      ctx.fillStyle="rgba(0,0,0,.55)"; ctx.fillRect(180,145,360,70);
      ctx.fillStyle="#fff"; ctx.font="bold 28px Arial"; ctx.textAlign="center";
      ctx.fillText("Out of shots — Reset to try again",360,190); ctx.textAlign="left";
    }
  }

  window.addEventListener("keydown", e => {
    if (e.key.toLowerCase()==="r") reset();
  });
  resetButton.addEventListener("click", reset);

  reset();
  requestAnimationFrame(update);
})();
</script>
"""))
```
